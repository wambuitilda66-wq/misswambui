```python
import pandas as pd

df = pd.read_csv('Sample - Superstore (1).csv', encoding='latin')
print("Loaded!")
print(df.shape)
df.head()
```

```python
# See all column names
print(df.columns.tolist())
print("\n")
#Check for missing values
print(df.isnull().sum())
```

```python
# CLEANING
# Convert dates
df['Order Date'] = pd.to_datetime(df['Order Date'],errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'],errors='coerce')
```

```python
#clean missing values
print("Before cleaning:")
print(df.isnull().sum())
```

```python
#clean missing values
print("\nAfter cleaning:")
print(df.isnull().sum())
print("\ncleaned! Ready for EDA")
```

```python
#EDA
print("--- BASIC STATS ---")
print(df.describe())

print("\n--- TOTAL SALES AND PROFIT ---")
print(f"Total Sales: ${df['Sales'].sum():,.2f}")
print(f"Total Profit: ${df['Profit'].sum():,.2f}")
print(f"Total Orders: {df['Order ID'].nunique()}")

print("\n--- Sales by Region ---")
print(df.groupby('Region')['Sales'].sum().sort_values(ascending=False))

print("\n---Sales by Category ---")
print(df.groupby('Category')['Sales'].sum().sort_values(ascending=False))

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sales Trends
df['Month-Year'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('Month-Year')['Sales'].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot()
plt.title('1. Sales Trends Over Time')
plt.xlabel('Month-Year')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Top Products
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_products.values, y=top_products.index, palette='viridis')
plt.title('2. Top 10 Products by Sales')
plt.xlabel('Total Sales ($)')
plt.tight_layout()
plt.show()

# Profit Distribution
plt.figure(figsize=(10,5))
sns.histplot(df['Profit'], bins=50, kde=True, color='green')
plt.title('3. Profit Distribution')
plt.xlabel('Profit ($)')
plt.tight_layout()
plt.show()

# Regional Comparison
regional = df.groupby('Region')[['Sales', 'Profit']].sum()
plt.figure(figsize=(10,5))
regional.plot(kind='bar')
plt.title('4. Regional Comparison')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
